```
public interface IInstallable
{
    void Install();
}


public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}");
        Console.WriteLine($"Цена: {Price}");
    }

    public override string ToString()
    {
        return $"{Name} - {Price} руб.";
    }
}


public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores)
        : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Ядер: {Cores}");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        return a.Name == b.Name &&
               a.Price == b.Price &&
               a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is CPU other)
            return this == other;

        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Cores);
    }
}


public class RAM : Component, IInstallable
{
    public int SizeGb { get; set; }

    public RAM(string name, decimal price, int sizeGb)
        : base(name, price)
    {
        SizeGb = sizeGb;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объем: {SizeGb} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Оперативная память {Name} установлена.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        return a.Name == b.Name &&
               a.Price == b.Price &&
               a.SizeGb == b.SizeGb;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is RAM other)
            return this == other;

        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, SizeGb);
    }
}


public class GPU : Component, IInstallable
{
    public int MemoryGb { get; set; }

    public GPU(string name, decimal price, int memoryGb)
        : base(name, price)
    {
        MemoryGb = memoryGb;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Видеопамять: {MemoryGb} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        return a.Name == b.Name &&
               a.Price == b.Price &&
               a.MemoryGb == b.MemoryGb;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is GPU other)
            return this == other;

        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, MemoryGb);
    }
}


public class HDD : Component, IInstallable
{
    public int CapacityGb { get; set; }

    public HDD(string name, decimal price, int capacityGb)
        : base(name, price)
    {
        CapacityGb = capacityGb;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объем: {CapacityGb} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Накопитель {Name} установлен.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        return a.Name == b.Name &&
               a.Price == b.Price &&
               a.CapacityGb == b.CapacityGb;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is HDD other)
            return this == other;

        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, CapacityGb);
    }
}


public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        components.Remove(component);
        Console.WriteLine($"Компонент {component.Name} удален.");
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;

        foreach (Component component in components)
        {
            total += component.Price;
        }

        return total;
    }

    public void ShowComponents()
    {
        foreach (Component component in components)
        {
            component.GetInfo();
            Console.WriteLine();
        }
    }
}


public static class ComputerFactory
{
    public static Computer CreateComputer<T1, T2>(
        T1 component1,
        T2 component2)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
    {
        Computer computer = new Computer();

        computer.AddComponent(component1);
        computer.AddComponent(component2);

        return computer;
    }
}


CPU cpu = new CPU(
    "Intel Core i7",
    30000m,
    8
);

RAM ram = new RAM(
    "Kingston Fury",
    8000m,
    16
);

GPU gpu = new GPU(
    "RTX 4070",
    70000m,
    12
);

HDD hdd = new HDD(
    "Seagate",
    6000m,
    1000
);


Computer computer = new Computer();

computer.AddComponent(cpu);
computer.AddComponent(ram);
computer.AddComponent(gpu);
computer.AddComponent(hdd);

Console.WriteLine();

computer.ShowComponents();

Console.WriteLine(
    $"Общая стоимость: {computer.GetTotalPrice()} руб."
);

Console.WriteLine();


CPU cpu2 = new CPU(
    "Intel Core i7",
    30000m,
    8
);

Console.WriteLine($"CPU одинаковые: {cpu == cpu2}");

Console.WriteLine();


computer.RemoveComponent(hdd);

Console.WriteLine();


Computer secondComputer =
    ComputerFactory.CreateComputer(
        new CPU("Ryzen 5", 20000m, 6),
        new RAM("Corsair", 7000m, 16)
    );

Console.WriteLine(
    $"Цена второго компьютера: " +
    $"{secondComputer.GetTotalPrice()} руб."
);
```
