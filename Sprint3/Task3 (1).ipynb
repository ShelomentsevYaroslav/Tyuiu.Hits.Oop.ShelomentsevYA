<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет собой систему управления задачами.

Базовый класс Task содержит общие данные задачи:
- идентификатор задачи;
- название;
- приоритет;
- исполнитель.

От класса Task наследуются:
- DelegateTask — делегированная задача;
- TeamTask — командная задача;
- ResearchTask — исследовательская задача.

Каждый производный класс добавляет собственные свойства и методы и переопределяет отдельные методы базового класса.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public string AssignedTo { get; set; }

    public Task(
        int taskId,
        string taskName,
        string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = "Не назначен";
    }

    public virtual void MarkAsComplete()
    {
        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена."
        );
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"ID: {TaskId}");
        Console.WriteLine($"Название: {TaskName}");
        Console.WriteLine($"Приоритет: {Priority}");
        Console.WriteLine($"Исполнитель: {AssignedTo}");
    }

    public virtual void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" назначена: {person}"
        );
    }

    public void ReassignTo(string person, string reason)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" назначена: {person}. " +
            $"Причина: {reason}"
        );
    }
}


public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string DelegatedBy { get; set; }

    public DelegateTask(
        int taskId,
        string taskName,
        string priority,
        DateTime dueDate,
        string delegatedBy)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
        DelegatedBy = delegatedBy;
    }

    public override void MarkAsComplete()
    {
        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена. " +
            $"Срок: {DueDate:d}"
        );
    }

    public void ShowDeadline()
    {
        Console.WriteLine(
            $"Срок выполнения: {DueDate:d}"
        );
    }
}


public class TeamTask : Task
{
    public string TeamName { get; set; }
    public int MemberCount { get; set; }

    public TeamTask(
        int taskId,
        string taskName,
        string priority,
        string teamName,
        int memberCount)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
        MemberCount = memberCount;
    }

    public override void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" команды " +
            $"\"{TeamName}\" назначена: {person}"
        );
    }

    public void AddMember()
    {
        MemberCount++;

        Console.WriteLine(
            $"Количество участников: {MemberCount}"
        );
    }
}


public class ResearchTask : Task
{
    public string DataSource { get; set; }
    public string ResearchMethod { get; set; }

    public ResearchTask(
        int taskId,
        string taskName,
        string priority,
        string dataSource,
        string researchMethod)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();

        Console.WriteLine(
            $"Источник данных: {DataSource}"
        );

        Console.WriteLine(
            $"Метод исследования: {ResearchMethod}"
        );
    }

    public void ChangeResearchMethod(string method)
    {
        ResearchMethod = method;

        Console.WriteLine(
            $"Новый метод исследования: {ResearchMethod}"
        );
    }
}


public class TaskCollection<T> where T : Task
{
    private List<T> tasks = new List<T>();

    public void Add(T task)
    {
        tasks.Add(task);
    }

    public void Remove(T task)
    {
        tasks.Remove(task);
    }

    public void ShowTasks()
    {
        foreach (T task in tasks)
        {
            Console.WriteLine("----------------");
            task.GetTaskDetails();
        }
    }
}


DelegateTask task1 = new DelegateTask(
    1,
    "Подготовить отчет",
    "Высокий",
    new DateTime(2026, 10, 30),
    "Руководитель"
);

TeamTask task2 = new TeamTask(
    2,
    "Разработать интерфейс",
    "Средний",
    "Команда разработки",
    5
);

ResearchTask task3 = new ResearchTask(
    3,
    "Провести исследование",
    "Высокий",
    "Научные статьи",
    "Сравнительный анализ"
);


List<Task> taskList = new List<Task>()
{
    task1,
    task2,
    task3
};

Console.WriteLine("ПОЛИМОРФИЗМ:");

foreach (Task task in taskList)
{
    task.GetTaskDetails();
    Console.WriteLine();
}


Console.WriteLine("ПЕРЕКРЫТИЕ:");

task1.MarkAsComplete();
task2.ReassignTo("Иван");
task3.GetTaskDetails();


Console.WriteLine();

Console.WriteLine("ПЕРЕГРУЗКА:");

task1.ReassignTo("Петр");

task1.ReassignTo(
    "Алексей",
    "Петр ушел в отпуск"
);


Console.WriteLine();

task1.ShowDeadline();
task2.AddMember();
task3.ChangeResearchMethod(
    "Статистический анализ"
);


Console.WriteLine();

Console.WriteLine("GENERIC:");

TaskCollection<Task> collection =
    new TaskCollection<Task>();

collection.Add(task1);
collection.Add(task2);
collection.Add(task3);

collection.ShowTasks();
```
