[Проект представляет систему управления задачами. Базовый класс Task содержит общие данные задачи: идентификатор, название, приоритет и исполнителя. На его основе создаются специализированные классы DelegateTask, TeamTask и ResearchTask. Каждый производный класс добавляет собственные свойства и переопределяет отдельные методы базового класса. В проекте используются конструкторы, свойства с геттерами и сеттерами, наследование, полиморфизм и взаимодействие объектов.]

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public class Task
{
    private int _taskId;
    private string _taskName;
    private string _priority;
    private string _assignedTo;

    public int TaskId
    {
        get { return _taskId; }
        set { _taskId = value; }
    }

    public string TaskName
    {
        get { return _taskName; }
        set { _taskName = value; }
    }

    public string Priority
    {
        get { return _priority; }
        set { _priority = value; }
    }

    public string AssignedTo
    {
        get { return _assignedTo; }
        set { _assignedTo = value; }
    }

    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = "Не назначен";
    }

    public virtual void MarkAsComplete()
    {
        Console.WriteLine($"Задача \"{TaskName}\" выполнена");
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"ID: {TaskId}");
        Console.WriteLine($"Название: {TaskName}");
        Console.WriteLine($"Приоритет: {Priority}");
        Console.WriteLine($"Исполнитель: {AssignedTo}");
    }

    public virtual void ReassignTo(string person)
    {
        AssignedTo = person;
        Console.WriteLine($"Задача \"{TaskName}\" назначена: {person}");
    }
}


public class DelegateTask : Task
{
    private DateTime _dueDate;

    public DateTime DueDate
    {
        get { return _dueDate; }
        set { _dueDate = value; }
    }

    public DelegateTask(
        int taskId,
        string taskName,
        string priority,
        DateTime dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    public override void MarkAsComplete()
    {
        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена. Дата выполнения: {DueDate:d}"
        );
    }
}


public class TeamTask : Task
{
    private string _teamName;

    public string TeamName
    {
        get { return _teamName; }
        set { _teamName = value; }
    }

    public TeamTask(
        int taskId,
        string taskName,
        string priority,
        string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }

    public override void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" команды \"{TeamName}\" переназначена: {person}"
        );
    }
}


public class ResearchTask : Task
{
    private string _dataSource;

    public string DataSource
    {
        get { return _dataSource; }
        set { _dataSource = value; }
    }

    public ResearchTask(
        int taskId,
        string taskName,
        string priority,
        string dataSource)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();

        Console.WriteLine($"Источник данных: {DataSource}");
    }
}


public class ProjectManager
{
    public string Name { get; set; }

    public ProjectManager(string name)
    {
        Name = name;
    }

    public void AssignTask(Task task, string person)
    {
        Console.WriteLine($"{Name} переназначает задачу");
        task.ReassignTo(person);
    }

    public void ShowTask(Task task)
    {
        task.GetTaskDetails();
    }
}
```

```
DelegateTask task1 = new DelegateTask(
    1,
    "Подготовить отчет",
    "Высокий",
    new DateTime(2026, 9, 30)
);

TeamTask task2 = new TeamTask(
    2,
    "Разработать интерфейс",
    "Средний",
    "Команда разработки"
);

ResearchTask task3 = new ResearchTask(
    3,
    "Провести исследование",
    "Высокий",
    "Научные статьи"
);

ProjectManager manager = new ProjectManager("Иван");


Console.WriteLine("1. Проверка DelegateTask");
task1.MarkAsComplete();

Console.WriteLine();

Console.WriteLine("2. Проверка TeamTask");
manager.AssignTask(task2, "Петр");

Console.WriteLine();

Console.WriteLine("3. Проверка ResearchTask");
manager.ShowTask(task3);
```
