```
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;

        Console.WriteLine(
            $"{Name} выросло на {growth} см. Новая высота: {Height} см."
        );
    }
}


public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(
        string name,
        float height,
        string color,
        int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine(
            $"Цветок {Name} распускается и радует глаз!"
        );
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}


public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(
        string name,
        float height,
        int age,
        string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine(
            $"Дерево {Name} покрывается новыми листьями!"
        );
    }

    public void GrowYears(int years)
    {
        Age += years;

        Console.WriteLine(
            $"Дерево {Name} стало старше. Новый возраст: {Age} лет."
        );
    }
}
```

```
Flower flower = new Flower(
    "Роза",
    35,
    "Красный",
    30
);

Tree tree = new Tree(
    "Береза",
    250,
    5,
    "Лиственные"
);


Console.WriteLine("Цветок:");

flower.Bloom();
flower.Grow(5);
flower.Wilt();

Console.WriteLine();


Console.WriteLine("Дерево:");

tree.Bloom();
tree.Grow(30);
tree.GrowYears(2);
```
