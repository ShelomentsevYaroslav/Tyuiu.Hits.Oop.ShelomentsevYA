[Проект представляет собой программу для учёта животных спасательного центра, их характеристик, состояния здоровья и подготовки к выполнению специальных задач.
Основой программы является класс Animal, содержащий общие характеристики животных. Производные классы Dog, Cat и Bird расширяют его функциональность. Класс RescueDog наследуется от Dog и предназначен для моделирования работы поисково-спасательных собак.
В программе реализованы три механизма наследования:
1. Простое наследование: Animal → Dog, Animal → Cat, Animal → Bird.
2. Сложное наследование: Animal → Dog → RescueDog.
3. Множественное наследование интерфейсов: ITrainable, ITrackable, IMedicalCare.
В рамках дополнительного задания классы расширены новыми свойствами и методами, позволяющими учитывать состояние животных, проводить тренировки, организовывать кормление и выполнять поисково-спасательные операции.]

```
public interface IDeadlineControlled
{
    void CheckDeadline();
}


public interface ITeamWork
{
    void ShowTeam();
}


public interface IResearchable
{
    void ShowDataSource();
}


public interface IReportable
{
    void CreateReport();
}



public class Task
{
    private int _taskId;
    private string _taskName;
    private string _priority;
    private string _assignedTo;
    private string _status;
    private string _description;

    public int TaskId
    {
        get { return _taskId; }
        set { _taskId = value; }
    }

    public string TaskName
    {
        get { return _taskName; }
        set { _taskName = value; }
    }

    public string Priority
    {
        get { return _priority; }
        set { _priority = value; }
    }

    public string AssignedTo
    {
        get { return _assignedTo; }
        set { _assignedTo = value; }
    }

    public string Status
    {
        get { return _status; }
        set { _status = value; }
    }

    public string Description
    {
        get { return _description; }
        set { _description = value; }
    }


    public Task(
        int taskId,
        string taskName,
        string priority,
        string description)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Description = description;

        AssignedTo = "Не назначен";
        Status = "Создана";
    }


    public virtual void MarkAsComplete()
    {
        Status = "Выполнена";

        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена"
        );
    }


    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"ID: {TaskId}");
        Console.WriteLine($"Название: {TaskName}");
        Console.WriteLine($"Приоритет: {Priority}");
        Console.WriteLine($"Исполнитель: {AssignedTo}");
        Console.WriteLine($"Статус: {Status}");
        Console.WriteLine($"Описание: {Description}");
    }


    public virtual void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" назначена: {person}"
        );
    }


    public void ChangePriority(string newPriority)
    {
        Priority = newPriority;

        Console.WriteLine(
            $"Приоритет задачи изменен на: {Priority}"
        );
    }


    public void StartTask()
    {
        Status = "В работе";

        Console.WriteLine(
            $"Задача \"{TaskName}\" взята в работу"
        );
    }
}



public class DelegateTask : Task, IDeadlineControlled
{
    private DateTime _dueDate;
    private string _delegatedBy;

    public DateTime DueDate
    {
        get { return _dueDate; }
        set { _dueDate = value; }
    }

    public string DelegatedBy
    {
        get { return _delegatedBy; }
        set { _delegatedBy = value; }
    }


    public DelegateTask(
        int taskId,
        string taskName,
        string priority,
        string description,
        DateTime dueDate,
        string delegatedBy)
        : base(
            taskId,
            taskName,
            priority,
            description)
    {
        DueDate = dueDate;
        DelegatedBy = delegatedBy;
    }


    public override void MarkAsComplete()
    {
        Status = "Выполнена";

        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена. " +
            $"Срок: {DueDate:d}"
        );
    }


    public void CheckDeadline()
    {
        Console.WriteLine(
            $"Срок задачи \"{TaskName}\": {DueDate:d}"
        );
    }


    public void ShowDelegator()
    {
        Console.WriteLine(
            $"Задачу поставил: {DelegatedBy}"
        );
    }
}



public class UrgentDelegateTask : DelegateTask
{
    public int UrgencyLevel { get; set; }
    public bool RequiresImmediateResponse { get; set; }


    public UrgentDelegateTask(
        int taskId,
        string taskName,
        string priority,
        string description,
        DateTime dueDate,
        string delegatedBy,
        int urgencyLevel,
        bool requiresImmediateResponse)
        : base(
            taskId,
            taskName,
            priority,
            description,
            dueDate,
            delegatedBy)
    {
        UrgencyLevel = urgencyLevel;
        RequiresImmediateResponse = requiresImmediateResponse;
    }


    public void ShowUrgency()
    {
        Console.WriteLine(
            $"Уровень срочности: {UrgencyLevel}"
        );

        Console.WriteLine(
            $"Требуется немедленная реакция: " +
            $"{RequiresImmediateResponse}"
        );
    }


    public override void GetTaskDetails()
    {
        base.GetTaskDetails();

        Console.WriteLine($"Срок: {DueDate:d}");
        Console.WriteLine($"Поставил: {DelegatedBy}");
        Console.WriteLine($"Срочность: {UrgencyLevel}");
    }
}



public class TeamTask : Task, ITeamWork, IReportable
{
    private string _teamName;
    private int _memberCount;

    public string TeamName
    {
        get { return _teamName; }
        set { _teamName = value; }
    }

    public int MemberCount
    {
        get { return _memberCount; }
        set { _memberCount = value; }
    }


    public TeamTask(
        int taskId,
        string taskName,
        string priority,
        string description,
        string teamName,
        int memberCount)
        : base(
            taskId,
            taskName,
            priority,
            description)
    {
        TeamName = teamName;
        MemberCount = memberCount;
    }


    public override void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" команды " +
            $"\"{TeamName}\" переназначена: {person}"
        );
    }


    public void ShowTeam()
    {
        Console.WriteLine(
            $"Команда: {TeamName}, " +
            $"участников: {MemberCount}"
        );
    }


    public void CreateReport()
    {
        Console.WriteLine(
            $"Отчет по командной задаче " +
            $"\"{TaskName}\" создан"
        );
    }


    public void AddMember()
    {
        MemberCount++;

        Console.WriteLine(
            $"В команде теперь {MemberCount} участников"
        );
    }
}



public class ResearchTask : Task, IResearchable, IReportable
{
    private string _dataSource;
    private string _researchMethod;

    public string DataSource
    {
        get { return _dataSource; }
        set { _dataSource = value; }
    }

    public string ResearchMethod
    {
        get { return _researchMethod; }
        set { _researchMethod = value; }
    }


    public ResearchTask(
        int taskId,
        string taskName,
        string priority,
        string description,
        string dataSource,
        string researchMethod)
        : base(
            taskId,
            taskName,
            priority,
            description)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
    }


    public override void GetTaskDetails()
    {
        base.GetTaskDetails();

        Console.WriteLine(
            $"Источник данных: {DataSource}"
        );

        Console.WriteLine(
            $"Метод исследования: {ResearchMethod}"
        );
    }


    public void ShowDataSource()
    {
        Console.WriteLine(
            $"Используемый источник: {DataSource}"
        );
    }


    public void CreateReport()
    {
        Console.WriteLine(
            $"Исследовательский отчет по задаче " +
            $"\"{TaskName}\" создан"
        );
    }


    public void ChangeResearchMethod(string newMethod)
    {
        ResearchMethod = newMethod;

        Console.WriteLine(
            $"Метод исследования изменен на: {ResearchMethod}"
        );
    }
}
```

```
DelegateTask task1 = new DelegateTask(
    1,
    "Подготовить отчет",
    "Высокий",
    "Подготовить итоговый отчет по проекту",
    new DateTime(2026, 10, 20),
    "Руководитель"
);


UrgentDelegateTask task2 = new UrgentDelegateTask(
    2,
    "Исправить критическую ошибку",
    "Критический",
    "Исправить ошибку перед выпуском системы",
    new DateTime(2026, 10, 12),
    "Руководитель проекта",
    5,
    true
);


TeamTask task3 = new TeamTask(
    3,
    "Разработать интерфейс",
    "Средний",
    "Создать интерфейс приложения",
    "Команда разработки",
    5
);


ResearchTask task4 = new ResearchTask(
    4,
    "Провести исследование",
    "Высокий",
    "Исследовать существующие решения",
    "Научные статьи",
    "Сравнительный анализ"
);



Console.WriteLine("1. ПРОСТОЕ НАСЛЕДОВАНИЕ");

task1.ReassignTo("Иван");
task1.CheckDeadline();
task1.ShowDelegator();
task1.MarkAsComplete();


Console.WriteLine();


Console.WriteLine("2. СЛОЖНОЕ НАСЛЕДОВАНИЕ");

task2.ReassignTo("Петр");
task2.ShowUrgency();
task2.GetTaskDetails();


Console.WriteLine();


Console.WriteLine("3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ");

task3.ShowTeam();
task3.CreateReport();
task3.AddMember();


Console.WriteLine();


Console.WriteLine("4. RESEARCH TASK");

task4.ShowDataSource();
task4.CreateReport();
task4.ChangeResearchMethod("Статистический анализ");
task4.GetTaskDetails();
```
