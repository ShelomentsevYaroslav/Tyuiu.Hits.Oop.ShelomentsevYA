<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет простую систему управления задачами. Базовый класс Task содержит общие данные задачи: идентификатор, название и приоритет, а также основные методы для завершения задачи, получения информации и переназначения исполнителя.
На основе базового класса созданы классы DelegateTask, TeamTask и ResearchTask. Каждый из них наследует общие свойства и методы класса Task, добавляет собственные данные и переопределяет отдельные методы. Это позволяет продемонстрировать наследование и полиморфизм в C#.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public string AssignedTo { get; set; }

    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = "Не назначен";
    }

    public virtual void MarkAsComplete()
    {
        Console.WriteLine($"Задача \"{TaskName}\" выполнена");
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"ID: {TaskId}");
        Console.WriteLine($"Название: {TaskName}");
        Console.WriteLine($"Приоритет: {Priority}");
        Console.WriteLine($"Исполнитель: {AssignedTo}");
    }

    public virtual void ReassignTo(string person)
    {
        AssignedTo = person;
        Console.WriteLine($"Задача \"{TaskName}\" назначена: {person}");
    }
}


// Делегатская задача
class DelegateTask : Task
{
    public DateTime DueDate { get; set; }

    public DelegateTask(
        int taskId,
        string taskName,
        string priority,
        DateTime dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    public override void MarkAsComplete()
    {
        Console.WriteLine(
            $"Задача \"{TaskName}\" выполнена. Дата выполнения: {DueDate:d}"
        );
    }
}


// Командная задача
class TeamTask : Task
{
    public string TeamName { get; set; }

    public TeamTask(
        int taskId,
        string taskName,
        string priority,
        string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }

    public override void ReassignTo(string person)
    {
        AssignedTo = person;

        Console.WriteLine(
            $"Задача \"{TaskName}\" команды \"{TeamName}\" переназначена: {person}"
        );
    }
}


// Исследовательская задача
class ResearchTask : Task
{
    public string DataSource { get; set; }

    public ResearchTask(
        int taskId,
        string taskName,
        string priority,
        string dataSource)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();

        Console.WriteLine($"Источник данных: {DataSource}");
    }
}
```

```
DelegateTask task1 = new DelegateTask(
    1,
    "Подготовить отчет",
    "Высокий",
    new DateTime(2026, 9, 20)
);

TeamTask task2 = new TeamTask(
    2,
    "Разработать интерфейс",
    "Средний",
    "Команда разработчиков"
);

ResearchTask task3 = new ResearchTask(
    3,
    "Провести исследование",
    "Высокий",
    "Научные статьи"
);


task1.MarkAsComplete();

Console.WriteLine();

task2.ReassignTo("Иван");

Console.WriteLine();

task3.GetTaskDetails();
```
