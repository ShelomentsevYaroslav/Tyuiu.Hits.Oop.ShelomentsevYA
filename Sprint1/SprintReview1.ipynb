```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;

    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }

        set
        {
            if (value >= 0)
                price = value;
        }
    }

    public int Quantity
    {
        get { return quantity; }

        set
        {
            if (value >= 0)
                quantity = value;
        }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(
        string name,
        decimal price,
        int quantity,
        string category,
        string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        int newQuantity = Quantity + amount;

        if (newQuantity >= 0)
        {
            Quantity = newQuantity;
        }
        else
        {
            Console.WriteLine(
                $"Недостаточно товара {Name} на складе."
            );
        }
    }

    public string GetProductInfo()
    {
        return
            $"Товар: {Name}, категория: {Category}, " +
            $"цена: {Price}, количество: {Quantity}, " +
            $"описание: {Description}";
    }
}


public class Seller
{
    private string name;
    private string employeeId;

    protected decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }

        set
        {
            if (value >= 0)
                salary = value;
        }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public IReadOnlyList<Product> Products
    {
        get { return products.AsReadOnly(); }
    }

    public Seller(
        string name,
        string employeeId,
        decimal salary,
        string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);

        Console.WriteLine(
            $"Продавец {Name} добавил товар {product.Name}."
        );
    }

    public void SellProduct(Product product, int quantity)
    {
        if (quantity <= 0)
        {
            Console.WriteLine("Количество должно быть больше нуля.");
            return;
        }

        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);

            Console.WriteLine(
                $"{Name} продал товар {product.Name} в количестве {quantity} шт."
            );
        }
        else
        {
            Console.WriteLine(
                $"Недостаточно товара {product.Name} для продажи."
            );
        }
    }

    public string GetSellerInfo()
    {
        return
            $"Продавец: {Name}, ID: {EmployeeId}, " +
            $"зарплата: {Salary}, контакты: {ContactInfo}";
    }
}


public class Store
{
    private string storeName;
    private string location;

    public string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(
        string storeName,
        string location,
        string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);

        Console.WriteLine(
            $"Продавец {seller.Name} добавлен в магазин {StoreName}."
        );
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары магазина {StoreName}:");

        foreach (Seller seller in sellers)
        {
            foreach (Product product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return
            $"Магазин: {StoreName}, адрес: {Location}, " +
            $"время работы: {storeHours}";
    }
}
```

```
Product product = new Product(
    "Ноутбук",
    80000m,
    10,
    "Электроника",
    "Игровой ноутбук"
);

Seller seller = new Seller(
    "Иван",
    "S001",
    50000m,
    "ivan@example.com"
);

Store store = new Store(
    "ТехноМаркет",
    "Тюмень",
    "09:00–21:00"
);


Console.WriteLine("1. Товар");

Console.WriteLine(product.GetProductInfo());

Console.WriteLine(
    $"Общая стоимость товара: {product.GetTotalPrice()}"
);


Console.WriteLine();

Console.WriteLine("2. Продавец");

seller.AddProduct(product);

Console.WriteLine(seller.GetSellerInfo());


Console.WriteLine();

Console.WriteLine("3. Продажа");

seller.SellProduct(product, 2);

Console.WriteLine(
    $"Осталось товара: {product.Quantity}"
);


Console.WriteLine();

Console.WriteLine("4. Магазин");

store.AddSeller(seller);

Console.WriteLine(store.GetStoreInfo());

store.ListProducts();
```
